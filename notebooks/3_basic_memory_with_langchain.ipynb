```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(
    model = "openai/gpt-oss-20b"
)
```

```python
res = llm.invoke("Hello, My name is Rohit Rout.")
print(res.content)
```

```python
res = llm.invoke("What is my name ?")
print(res.content)
```

```python
prompts = [
    {"role":"user", "content":"What is my name ?."},
    {"role":"ai", "content":"Hello Rohit! 👋 How can I assist you today?"},
    {"role":"user", "content":"What is my name ?"}
]
```

```python
res = llm.invoke(prompts)
print(res.content)
```

### Manage History

```python
history = []
while True:
    query = input("User: ")
    if query.lower() in ["exit", "quit", "bye"]:
        print("GoodBye 👋")
        break

    history.append({"role":"user", "content":query})
    print("User: ", query)

    res = llm.invoke(history)
    history.append({"role":"ai", "content":res.content})

    print(f"AI: {res.content}\n")
```
