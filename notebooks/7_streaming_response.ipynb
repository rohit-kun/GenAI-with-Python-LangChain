```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(
    model = "openai/gpt-oss-20b",
    streaming = True
)
```

```python
question = "Can you explain me GEN-AI ?"
```

```python
response = llm.stream(question)
for chunk in response:
    print(chunk.content, end=" ")
```
