```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
texts = """My dear friends, students, and respected farmers,

Today, I want to speak from the heart about three pillars that define the future of our nation:
our youth, our education system, and our farmers"""
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 50, chunk_overlap = 20)
```

```python
chunks = splitter.split_text(texts)
```
