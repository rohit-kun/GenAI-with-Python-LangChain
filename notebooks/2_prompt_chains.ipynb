### Prompt Templates - Dynamic Prompts

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


```

```python
llm = ChatGroq(
    model = "openai/gpt-oss-20b",
)

out = StrOutputParser()
```

```python
### My Self Chain Node

def transform_case(result:str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are a translator and translate input in {language}"},
    {"role":"user", "content":"{query}"},
])
```

```python
final_prompts = prompts.format_messages(language = "Hinglish",
                                         query = "I love Python, and JavaScript")

final_prompts
```

```python
res = llm.invoke(final_prompts)
print(res.content)
```

### Chains - Runnable

```python
prompts.invoke({"language":"Hindi", "query":"I love you."})
```

```python
res = llm.invoke(prompts.invoke({"language":"Hindi", "query":"I love you."}))
print(res.content)
```

```python
chains = prompts | llm

chains
```

```python
res = chains.invoke({"language":"hindi", "query":"I love LangChain."})
print(res.content)
```

```python
chains = prompts | llm | out
```

```python
res = chains.invoke({"language":"hindi", "query":"I love LangChain."})
print(res)
```

```python
chains = prompts | llm | out | transform_case
```

```python
res = chains.invoke({"language":"hindi", "query":"I love LangChain."})
print(res)
```
